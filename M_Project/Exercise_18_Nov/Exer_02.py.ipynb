```python
from pyspark.sql import SparkSession, functions as F
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, IntegerType, LongType, StringType, FloatType


# Initialize spark context
spark = SparkSession.builder.getOrCreate()

schema_Cust = StructType([StructField("Name", StringType(), True),
                          StructField("Age",IntegerType(), True),
                        ])

df = spark.read.csv("dbfs:/Volumes/sparkcatalog/default/practical/Data.csv", header=True, schema=schema_Cust,sep=',').createOrReplaceTempView("DataTable")


spark.sql("SELECT Upper(Name), Age FROM DataTable").show(2)


#df.write.saveAsTable("Data18_TA")
```
