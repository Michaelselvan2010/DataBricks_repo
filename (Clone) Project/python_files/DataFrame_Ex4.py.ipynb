```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, IntegerType

# Create a Spark session
spark = SparkSession.builder.appName("CustomSchemaExample").getOrCreate()

# Define schema manually
custom_schema = StructType([
    StructField("name", StringType(), True),
    StructField("age", IntegerType(), True)
])

# Read a CSV file with the specified schema
df = spark.read.csv("dbfs:/Volumes/sparkcatalog/default/practical/Data.csv", header=True, schema=custom_schema)

# Print schema
df.printSchema()

# Show data
df.show()
```
